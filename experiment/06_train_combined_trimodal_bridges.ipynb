# 06 — Train COMBINED trimodal bridges

Train genuine triplets, an architecture-matched cross-patient shuffled control, and a lab-missingness-only
control across three seeds. Completed runs are skipped; incompatible run protocols fail rather than overwrite.

```python
from pathlib import Path
import os, sys
CODE_ROOT = Path(os.environ.get('EXPERIMENT4_CODE_ROOT', str(Path.cwd())))
if not (CODE_ROOT / 'e4_common.py').is_file() and 'EXPERIMENT4_CODE_ROOT' not in os.environ:
    CODE_ROOT = Path.cwd()
if not CODE_ROOT.is_absolute() or not (CODE_ROOT / 'e4_common.py').is_file():
    raise FileNotFoundError('Set EXPERIMENT4_CODE_ROOT to the uploaded Experiment 4 source directory')
if str(CODE_ROOT) not in sys.path: sys.path.insert(0, str(CODE_ROOT))
from e4_common import ARTIFACT_ROOT, EXPERIMENT4_ROOT, SYMILE_ROOT
print({'code_root': str(CODE_ROOT), 'artifact_root': str(ARTIFACT_ROOT),
       'experiment4_root': str(EXPERIMENT4_ROOT), 'symile_root': str(SYMILE_ROOT)})
```

```python
import os
from e4_bridge import train
method = 'combined'
seeds = [int(x) for x in os.environ.get('E4_SEEDS', '43,44,45').split(',')]
epochs = int(os.environ.get('E4_EPOCHS', '50'))
batch_size = int(os.environ.get('E4_BATCH_SIZE', '128'))
runs = []
for seed in seeds:
    runs.append(train(method=method, pairing='genuine', labs_mode='values', seed=seed,
                      epochs=epochs, batch_size=batch_size))
    runs.append(train(method=method, pairing='shuffled', labs_mode='values', seed=seed,
                      epochs=epochs, batch_size=batch_size))
    runs.append(train(method=method, pairing='genuine', labs_mode='mask_only', seed=seed,
                      epochs=epochs, batch_size=batch_size))
    runs.append(train(method=method, pairing='genuine', labs_mode='values',
                      no_code_match=True, seed=seed, epochs=epochs, batch_size=batch_size))

runs
```
